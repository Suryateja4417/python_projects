```python
import pandas as pd
ft = pd.read_csv('stockerbot-export.csv',on_bad_lines='skip')
ft.head()
```

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')
nltk.download('subjectivity')
nltk.download('movie_review')
sia = SentimentIntensityAnalyzer()
```

```python
ft.describe()
```

```python
ft['symbols'].unique()
```

```python
ft['symbols'].value_counts()
```

```python
np = ft[ft['symbols'] == 'NFLX']
np_sort = np.sort_values(by ='timestamp')
```

```python
compound_scores = []

for index, row in np_sort.iterrows():
    tweet = row['text']
    analysis = sia.polarity_scores(tweet)
    compound_scores.append(analysis['compound'])

np_sort['sentimental_scores'] = compound_scores
np_sort
```

```python
import matplotlib.pyplot as plt
plt.plot(np_sort['timestamp'], compound_scores)
plt.ylabel('compound score')
plt.show()
```
